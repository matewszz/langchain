```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain.prompts import PromptTemplate

openai = ChatOpenAI(model_name='gpt-4o-mini') | StrOutputParser()
```

```python
prompt_template = PromptTemplate.from_template("Descreva as tendências do langChain em {ano}.")
runnable_sequence = prompt_template | openai
output = runnable_sequence.invoke({"ano": "2026"})
```

```python
print("Output:\n", output)
```
